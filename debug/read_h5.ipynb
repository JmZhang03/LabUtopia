```python
import h5py
import numpy as np
import matplotlib.pyplot as plt
import os
```

```python
h5_path = "../outputs/collect/All/Level1_pick/dataset/episode_0000.h5"
h5_file = h5py.File(h5_path, 'r')
keys = list(h5_file.keys())
print(f"Keys: {keys}")
for k in keys:
    print(f"key: {k}, value shape: {h5_file[k].shape}")
```

```python
actions = h5_file["actions"][:]
print("actions shape:", actions.shape)
print("actions mean:", actions.mean())
print("actions std:", actions.std())
print("actions min/max:", actions.min(), actions.max())

poses = h5_file["agent_pose"][:]
print("actions:", actions.shape, actions.mean(), actions.std())
print("poses:", poses.shape, poses.mean(), poses.std())
```

```python
plt.figure()
for i in range(actions.shape[1]):
    plt.plot(actions[:, i], label=f"dim {i}")
plt.legend()
plt.title("Action trajectory")
plt.show()
```

```python
img = h5_file["camera_1_rgb"][0]
img = img.transpose(1, 2, 0)
plt.imshow(img)
plt.title("camera_1_rgb frame 0")
plt.show()
```
