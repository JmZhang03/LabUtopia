```python
import torch
import numpy as np
```

```python
ckpt_path = "../outputs/train/2026.04.14/22.53.07_train_act_image_pick_open/checkpoints/epoch=060-train_loss=0.020.ckpt"

ckpt = torch.load(ckpt_path, map_location="cpu", weights_only=False)

print(type(ckpt))
print(ckpt.keys())
```

```python
for k, v in ckpt.items():
    if isinstance(v, dict):
        print(k, "=> dict size:", len(v))
    else:
        print(k, type(v))
```

```python
state_dict = ckpt.get("state_dicts", None)

if isinstance(state_dict, dict):
    print("num keys:", len(state_dict))
    print("sample keys:", list(state_dict.keys()))
```

```python
model = state_dict.get("model", None)
if isinstance(model, dict):
    keys = list(model.keys())
    print("num keys:", len(keys))
    print("contains transformer:", any("transformer" in k for k in keys))
    print("contains encoder:", any("encoder" in k for k in keys))
    print("contains actor:", any("actor" in k for k in keys))
```

```python
for k,v in model.items():
    print(f"Key: {k}, Value Shape: {v.shape}")
```
